```python
from env import config, BASE_DIR
from sqlalchemy import create_engine, text
```

```python
DATABASE_URL = config("HOUSING_DATABASE_URL", cast=str, default=None)
DATABASE_URL is not None
```

```python
engine = create_engine(str(DATABASE_URL))
```

```python
DATASET_DIR = BASE_DIR / "datasets"
DATASET_DIR.exists()
```

```python
HOUSE_CSV = DATASET_DIR / "housing_ds.csv"
HOUSE_CSV.exists()
```

```python
import pandas as pd
```

```python
df = pd.read_csv(str(HOUSE_CSV))
df.head(n=10)
```

```python
df.to_sql("kaggle_house_data", engine, if_exists='replace', index=False)
```
